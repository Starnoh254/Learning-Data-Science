```python
import urllib.request
import json

# This is a public API endpoint on the internet
url = "https://api.adviceslip.com/advice"

try : 
    # Your program is making an HTTP request to the internet right here:
    response = urllib.request.urlopen(url)
    data = json.loads(response.read())

    # Print out the piece of advice the internet gave back to us!
    print("✨ Advice from the internet: " + data['slip']['advice'])

except Exception as e:
    print("An error occurred while trying to fetch advice from the internet.")
    print("Error details:", e)


```
